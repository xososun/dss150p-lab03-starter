```python
import pandas as pd, json
c = pd.read_csv('data/source/customers.csv')
o = pd.read_csv('data/source/orders.csv')
p = pd.json_normalize(json.load(open('data/source/products.json', encoding='utf-8')))
for name, df in [('customers', c), ('products', p), ('orders', o)]:
    print(f'\n== {name}: {df.shape}'); print(df.dtypes.to_string()); print(df.isna().sum().to_string())
print('\ndup customer_id:', c.customer_id.duplicated().sum())
print('dup product_id:', p.product_id.duplicated().sum())
print('dup order_id:', o.order_id.duplicated().sum())
print(o.status.value_counts(dropna=False).to_string())
print(o.quantity.describe().to_string())
print(p.unit_price.astype(str).sort_values().head(3).to_string(), p.unit_price.astype(str).sort_values().tail(3).to_string(), sep='\n')
```
